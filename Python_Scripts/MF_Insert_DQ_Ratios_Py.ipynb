```python
import pandas as pd
import psycopg2
from datetime import datetime


df = pd.read_excel(r'C:\Users\H59045\Downloads\SF_MultiFamily Loans.xlsx')
```

```python


print(df.head())
```

```python
conn = psycopg2.connect(
    host = 'localhost',
    database='postgres',
    user='postgres',
    password=input()
)

cur = conn.cursor()

insert_query = """ INSERT INTO gnma_mbs_mf_loans (CUSIP,Pool_Number,Pool_Indicator,Pool_Type,Security_Interest_Rate,Issuer_Number,Loan_Type,Loan_Interest_Rate,Unpaid_Principal_Balance,Months_Delinquent,Property_Name,Property_Street,Property_City,Property_State,Property_Zip,Report_Date) VALUES (%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s)"""

for index, row in df.iterrows():
    cur.execute(insert_query, (
        row['CUSIP'],
        row['Pool_Number'],
        row['Pool_Indicator'],
        row['Pool_Type'],
        row['Security_Interest_Rate'],
        row['Issuer_Number'],
        row['Loan_Type'],
        row['Loan_Interest_Rate'],
        row['Unpaid_Principal_Balance'],
        row['Months_Delinquent'],
        row['Property_Name'],
        row['Property_Street'],
        row['Property_City'],
        row['Property_State'],
        row['Property_Zip'],
        row['Report_Date'],
    ))


conn.commit()
cur.close()
conn.close()

print(f"Data inserted successfully")
```
