```python
import psycopg2.extras

df['As_of_Date'] = input('As of Date')
df= df.where(pd.notna(df), None)

conn = psycopg2.connect(dbname='postgres', user='postgres', password=input(), host='localhost') 
cursor = conn.cursor()

# Insert dataframe
for index, row in df.iterrows():
    try:
        cursor.execute("""
            INSERT INTO sf_new_issuance_loan_level VALUES (%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s,%s)
        """, tuple(row))
    except Exception as e:
        print(f"Error on row {index}: {e}")
        print(f"Row values: {tuple(row)}")
        print(f"Row length: {len(tuple(row))}")
        break

conn.commit()

print(f"{len(df)} loans inserted into new Issuances")

```
