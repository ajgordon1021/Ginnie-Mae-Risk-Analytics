```python
import pandas as pd
import psycopg2
from datetime import datetime
from decimal import Decimal

df = pd.read_excel(r' Put file name here')
```

```python
df['FIC'] = df['FIC'].apply(lambda x: Decimal(str(x)))

print(df.head())
```

```python
conn = psycopg2.connect(
    host = 'localhost',
    database='postgres',
    user='postgres',
    password=input()
)

cur = conn.cursor()

insert_query = """ INSERT INTO sf_gnma_dq_ratios (ID, Report_Date,DQ2,DQ3,DQP,FIC) VALUES (%s,%s,%s,%s,%s,%s)"""

for index, row in df.iterrows():
    cur.execute(insert_query, (
        row['ID'],
        row['Report Date'],
        row['DQ2'],
        row['DQ3'],
        row['DQP'],
        Decimal(str(row['FIC']))
    ))


conn.commit()
cur.close()
conn.close()

print(f"Data inserted successfully")
```
